````python
from typing import TypedDict, List, Optional, Literal, Dict
import time
import json
import re
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.runnables import RunnablePassthrough
from langgraph.graph import StateGraph

# Типы для состояния
class DocumentState(TypedDict):
    text: str
    extracted_data: Dict[str, List[str]]  # Словарь с извлеченными данными по атрибутам
    validation_results: Dict[str, dict]  # Результаты валидации по атрибутам
    hints: Dict[str, str]  # Подсказки по атрибутам
    attempts: int
    status: Literal["pending", "success", "max_attempts"]
    result: str

# Конфигурация атрибутов для извлечения
ATTRIBUTES_CONFIG = {
    "product_names": {
        "system_prompt": "Извлеки все названия товаров из текста.",
        "human_prompt": "Верни только список названий товаров через запятую."
    },
    "supplier_info": {
        "system_prompt": "Извлеки информацию о поставщике: название компании, контактные данные.",
        "human_prompt": "Верни только список элементов информации о поставщике через запятую."
    },
    "storage_conditions": {
        "system_prompt": "Извлеки требования и условия хранения товаров.",
        "human_prompt": "Верни только список условий хранения через запятую."
    }
}

# Создаем граф
workflow = StateGraph(DocumentState)

# Нода 1: Извлечение всех атрибутов
def extract_attributes(state: DocumentState):
    time.sleep(1.1)
    extracted_data = state.get("extracted_data", {})
    
    for attr, config in ATTRIBUTES_CONFIG.items():
        time.sleep(2.1)
        prompt = ChatPromptTemplate.from_messages([
            ("system", "{system_promt}  Учитывай подсказки из предыдущих попыток."),
            ("human", 
             """Текст: {text}\n\n
             Подсказка:  {hint}\n
             {human_promt}""")
        ])

        hint = [state['hints'].get(attr, '') if state.get('hints') else '']

        
        items = (prompt | llm).invoke({
            'system_promt' : config["system_prompt"],
            'text' : state['text'],
            'hint' : hint,
            'human_promt' : config['human_prompt']
        }).content
        print('items')
        extracted_data[attr] = [item.strip() for item in items.split(",")]
    
    return {
        "extracted_data": extracted_data,
        "attempts": state.get("attempts", 0) + 1
    }

# Нода 2: Валидация извлеченных данных
def validate_attributes(state: DocumentState):
    time.sleep(1.1)
    validation_results = {}
    
    for attr, config in ATTRIBUTES_CONFIG.items():
        time.sleep(2.1)
        prompt = ChatPromptTemplate.from_messages([
            ("system", f"Проверь полноту извлечения {attr}."),
            ("human", 
             """Текст: {text}\n\n
             Извлеченные данные: {items}\n\n
             Ответь в JSON:\n
             {{\n
               "all_found": bool,\n
               "missing": [str],\n
               "hint": str\n
             }}""")
        ])
        
        validation = (prompt | llm)
        val_result = validation.invoke({
            'text': state['text'],
            'items': ', '.join(state['extracted_data'].get(attr, []))
        })
        print('valid')
        response = val_result.content 
        json_match = re.search(r'```json\n({.*?})\n```', response, re.DOTALL) or re.search(r'\{.*\}', response)
        if json_match:
            try:
                validation_data = json.loads(json_match.group(1))
                validation_results[attr] = {
                    "all_found": validation_data.get("all_found", False),
                    "missing": validation_data.get("missing", []),
                    "hint": validation_data.get("hint", "Нет подсказки")
                }
            except json.JSONDecodeError:
                validation_results[attr] = {
                    "all_found": False,
                    "missing": ["Не удалось распарсить ответ"],
                    "hint": "Попробуйте уточнить запрос"
                }
    
    return {"validation_results": validation_results}

# Нода 3: Проверка условий завершения
def check_termination(state: DocumentState):
    all_valid = all(
        result.get("all_found", False)
        for result in state["validation_results"].values()
    )
    
    if all_valid:
        return {"status": "success"}
    elif state["attempts"] >= 3:
        return {"status": "max_attempts"}
    return {"status": "pending"}

# Нода 4: Генерация подсказок
def generate_hints(state: DocumentState):
    hints = {}
    for attr, result in state["validation_results"].items():
        hints[attr] = result.get("hint", "Нет подсказки")
    return {"hints": hints}

# Нода 5: Формирование финального результата
def format_result(state: DocumentState):
    if state["status"] == "success":
        result = "Успешно извлечены данные:\n"
        for attr, items in state["extracted_data"].items():
            result += f"- {attr}: {items}\n"
    else:
        result = "Лучший результат после 3 попыток:\n"
        for attr, items in state["extracted_data"].items():
            result += f"- {attr}: {items}\n"
    
    return {"result": result}

# Добавляем ноды в граф
workflow.add_node("extract", extract_attributes)
workflow.add_node("validate", validate_attributes)
workflow.add_node("check_termination", check_termination)
workflow.add_node("generate_hints", generate_hints)
workflow.add_node("format_result", format_result)

# Определяем потоки
workflow.set_entry_point("extract")
workflow.add_edge("extract", "validate")
workflow.add_edge("validate", "check_termination")

# Условные переходы
workflow.add_conditional_edges(
    "check_termination",
    lambda state: state["status"],
    {
        "pending": "generate_hints",
        "success": "format_result",
        "max_attempts": "format_result"
    }
)

workflow.add_edge("generate_hints", "extract")  # Повторная попытка

# Точки выхода
workflow.set_finish_point("format_result")

# Компилируем граф
app = workflow.compile()

# Функция для обработки нескольких документов
def process_documents(documents: List[str], attributes_config: Dict = None):
    global ATTRIBUTES_CONFIG
    if attributes_config:
        ATTRIBUTES_CONFIG = attributes_config
    
    full_results = []
    results = []
    for doc in documents:
        state = {
            "text": doc,
            "extracted_data": {},
            "validation_results": {},
            "hints": {},
            "attempts": 0,
            "status": "pending",
            "result": ""
        }
        result = app.invoke(state)
        results.append(result["result"])
        full_results.append(result)
    
    return results , full_results
````

````python
from docx import Document

def read_docx(file_path):
    """
    Read text content from a Word document (.docx)
    
    Args:
        file_path (str): Path to the .docx file
        
    Returns:
        str: All text content from the document
    """
    doc = Document(file_path)
    full_text = []
    
    for paragraph in doc.paragraphs:
        full_text.append(paragraph.text)
    
    return '\n'.join(full_text)

# Example usage
file_path = '/home/rus/vs_folder/data/ТЗ_Описание объекта закупки.docx'
content = read_docx(file_path)
print(content)
````

````python
import json
from datetime import datetime
import os

def save_experiment_results(documents, config, results, full_results, experiment_name=None):
    """
    Сохраняет результаты экспериментов в структурированном виде
    
    Args:
        documents: Список документов
        config: Использованная конфигурация
        results: Основные результаты
        full_results: Полные результаты
        experiment_name: Название эксперимента (опционально)
    """
    # Создаем папку для экспериментов, если ее нет
    exp_dir_path = '../data/'
    os.makedirs(f"{exp_dir_path}experiments", exist_ok=True)
    
    # Генерируем имя файла
    timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")
    experiment_name = experiment_name or f"exp_{timestamp}"
    filename = f"{exp_dir_path}experiments/{experiment_name}.json"
    
    # Подготавливаем данные для сохранения
    experiment_data = {
        "timestamp": timestamp,
        "experiment_name": experiment_name,
        "config": config,
        "documents": [str(doc) for doc in documents],  # Преобразуем документы в строки
        "results": results,
        "full_results": full_results
    }
    
    # Сохраняем в JSON
    with open(filename, "w", encoding="utf-8") as f:
        json.dump(experiment_data, f, ensure_ascii=False, indent=2)
    
    return filename

def load_experiment_results(filename):
    """Загружает сохраненные результаты эксперимента"""
    with open(filename, "r", encoding="utf-8") as f:
        return json.load(f)

def compare_experiments(experiment_files):
    """Сравнивает несколько экспериментов"""
    comparisons = []
    
    for file in experiment_files:
        data = load_experiment_results(file)
        comparisons.append({
            "experiment": data["experiment_name"],
            "config": data["config"],
            "summary": {k: len(v) if isinstance(v, list) else v for k, v in data["results"].items()}
        })
    
    return comparisons

````
